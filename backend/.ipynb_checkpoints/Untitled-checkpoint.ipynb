```python
from typing import Literal, Annotated
from pydantic import BaseModel ,Field

from pydantic import BaseModel, Field, ConfigDict
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder


class General(BaseModel):
    """Route requests that require greeting , general question answer."""
    model_config = ConfigDict(extra="forbid")
    agent: Literal["general"] = "gerenal"

class END(BaseModel):
    """Route requests that if task is complate or input query is empty"""
    model_config = ConfigDict(extra="forbid")
    agent: Literal["end"] = "end"

class RAG(BaseModel):
    """Route requests that require knowledge/information retrieval."""
    model_config = ConfigDict(extra="forbid")
    agent: Literal["rag"] = "rag"
    query: str = Field(...,min_length=1,description=("The user's information-retrieval question rewritten as a ""clear standalone query."),)



class MCP(BaseModel):
    """Route requests that require an MCP/tool operation."""
    model_config = ConfigDict(extra="forbid")
    agent: Literal["mcp"] = "mcp"
    query: str = Field(...,min_length=1,description=("Natural-language description of the action that ""the MCP worker needs to perform."))


class Schedule(BaseModel):
    """Route requests related to scheduling."""
    model_config = ConfigDict(extra="forbid")
    agent: Literal["schedule"] = "schedule"
    type: Literal["cron", "interval", "date"] = Field(...,description=("Scheduling strategy. ""'date' for a one-time execution, ""'cron' for recurring calendar-based execution, ""'interval' for recurring execution at a fixed interval."),)
    dateTime: str = Field(...,min_length=1,description=("The scheduling expression or datetime required by the ""selected schedule type."),)


class SupervisorResponse(BaseModel):
    """Decision produced by the supervisor."""
    model_config = ConfigDict(extra="forbid")
    response: Annotated[RAG | MCP|END | General | Schedule,Field(description="Exactly one routing decision.")]
    plan:

SUPERVISOR_SYSTEM_PROMPT = """
You are the supervisor/router of an AI agent system.
Your ONLY responsibility is to determine which agent should handle
the user's request.

Available agents:
1. rag
   Use when the user is asking for information, explanation,
   knowledge retrieval, document retrieval, search, or questions
   that can be answered from a knowledge/RAG system.
2. MCP
   Use when the request requires an external tool, API, database, customer relation managment
   filesystem, service, or other MCP capability.
3. schedule
   Use when the user wants to create, modify, cancel, inspect,
   or otherwise manage a scheduled task.
3. General
   Use when the user wants to create, modify, cancel, inspect,
   or otherwise manage a scheduled task.
3. end
   Use when the user wants to create, modify, cancel, inspect,
   or otherwise manage a scheduled task.

ROUTING RULES:
- Choose exactly ONE agent.
- Do not execute the user's request.
- Do not answer the user's question.
- Do not invent information that is not present in the conversation.
- If the request is primarily about scheduling, choose schedule.
- If the request requires an external tool/action, choose MCP.
- If the request is primarily an information/knowledge question,
  choose rag.
- For RAG, rewrite the user's request into a concise standalone query.
- For MCP, provide the MCP server identifier/URL only when it is
  known from the conversation or system context. Never invent a URL.
- For schedule, determine whether the schedule is:
    - date: one-time execution
    - cron: recurring cron-style schedule
    - interval: recurring fixed-duration interval
- Return only the structured response required by the schema.
"""

prompt = ChatPromptTemplate.from_messages([("system", SUPERVISOR_SYSTEM_PROMPT),MessagesPlaceholder(variable_name="messages"),])
supervisor_llm = prompt | llm.with_structured_output(SupervisorResponse)
```

```python
from typing import Literal
from pydantic import BaseModel, Field

class PlanStep(BaseModel):
    agent: Literal["crm","email","rag","MCP","schedule","general"]
    task: str = Field(description="Specific task the worker must execute.")


class Plan(BaseModel):
    steps: list[PlanStep] = Field(min_length=1,description="Ordered execution steps.")


from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
planner_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are a Planning Agent.

Your ONLY job is to create an ordered execution plan
for the user's request.

Available workers and their capabilities:

General worker:
Can greet user,general question answer, summary .

CRM worker:
Can retrieve invoices, customers, orders, and CRM records.

RAG worker:
Can search documents and knowledge bases.

MCP worker:
Can perform external tool/API operations.

Email worker:
Can send emails and attachments.

Schedule worker:
Can create, update, and cancel schedules.

IMPORTANT:

The capability descriptions above are NOT tasks.
Never copy capability descriptions into the plan.

Planning rules:

1. Create only tasks that are explicitly required to accomplish
   the user's request.

2. Each plan step must represent ONE concrete action.

3. Assign exactly ONE worker to each step.

4. If a task requires information that is not currently available,
   create a prerequisite step to obtain that information.

5. Preserve the correct execution order.

6. Do not create generic tasks such as:
   - "Call APIs"
   - "Execute external tools"
   - "Perform external actions"
   - "Search capabilities"

7. Do not repeat the same task.

8. Do not execute any task.

9. Do not explain the plan.

10. Return ONLY the structured Plan object.

Example:

User:
"Find my invoice and send it by email."

Correct plan:

Step 1:
agent = CRM
task = "Find the user's invoice."

Step 2:
agent = email
task = "Send the retrieved invoice by email."

Incorrect plan:

Step 1:
agent = MCP
task = "Call APIs"

Step 2:
agent = MCP
task = "Execute external tools"

Step 3:
agent = MCP
task = "Perform external actions"

Those are capabilities, not user tasks.
"""
    ),
    MessagesPlaceholder("messages"),
])

planner_llm = (
    planner_prompt
    | llm.with_structured_output(Plan)
)

from typing import Annotated
from typing_extensions import TypedDict

from langchain_core.messages import BaseMessage
from langgraph.graph.message import add_messages


class AgentState(TypedDict, total=False):
    messages: Annotated[list[BaseMessage], add_messages]
    plan: Plan
    current_step: int
    results: dict[int, str]
    final_response: str

from typing import Literal
from pydantic import BaseModel, Field, ConfigDict


class SupervisorResponse(BaseModel):
    model_config = ConfigDict(extra="forbid")
    next: Literal["planner","rag","MCP","schedule","email","finish",] = Field( description="Route actionable user requests to planner.")


supervisor_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are the Supervisor Agent.

Your responsibility is only to decide whether the user's
request requires planning.

Rules:

1. If the user asks you to perform any task, route to "planner".
2. Do NOT route directly to workers.
3. Do NOT select RAG, MCP, CRM, email, or schedule workers.
4. The Planner is responsible for breaking the request into steps.
5. Use "finish" only when there is nothing that needs to be executed.
"""
    ),
    MessagesPlaceholder("messages"),
])

supervisor_llm = (
    supervisor_prompt
    | llm.with_structured_output(SupervisorResponse)
)


def planner_agent(state: AgentState):
    plan = planner_llm.invoke({"messages": state["messages"]})
    return {"plan": plan,"current_step": 0,"results": {}}

```

```python
from typing import Annotated, Literal

from pydantic import BaseModel, Field, ConfigDict
from typing_extensions import TypedDict

from langchain_core.messages import BaseMessage, HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages


# ============================================================
# SUPERVISOR
# ============================================================

class SupervisorResponse(BaseModel):
    model_config = ConfigDict(extra="forbid")

    next: Literal["planner", "finish"] = Field(
        description=(
            "Route actionable user requests to planner. "
            "Use finish only when no task needs to be executed."
        )
    )


# ============================================================
# PLANNER
# ============================================================

class PlanStep(BaseModel):
    model_config = ConfigDict(extra="forbid")

    agent: Literal[
        "CRM",
        "rag",
        "MCP",
        "schedule",
        "email",
    ]

    task: str = Field(
        min_length=1,
        description="One concrete task for the selected worker."
    )


class Plan(BaseModel):
    model_config = ConfigDict(extra="forbid")

    steps: list[PlanStep] = Field(
        min_length=1,
        max_length=10,
        description="Ordered execution steps."
    )


# ============================================================
# STATE
# ============================================================

class AgentState(TypedDict, total=False):

    messages: Annotated[
        list[BaseMessage],
        add_messages
    ]

    plan: Plan
    next:Literal["planner","finish"]
    current_step: int

    results: dict[int, str] # for dynamic

    final_response: str

supervisor_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are the Supervisor Agent.

Your job is to decide whether the user's request
needs to be planned.

Rules:

1. If the user asks to perform any task, return "planner".
2. Do not directly select workers.
3. Do not execute tasks.
4. Do not choose CRM, RAG, MCP, email, or schedule.
5. Use "finish" only when no task needs to be executed.

Return only the structured response.
"""
    ),
    MessagesPlaceholder("messages"),
])


supervisor_llm = (
    supervisor_prompt
    | llm.with_structured_output(SupervisorResponse)
)
async def supervisor_node(state: AgentState):

    response = await supervisor_llm.ainvoke({
        "messages": state["messages"]
    })

    return {
        "next": response.next
    }


planner_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are a Planning Agent.

Convert the user's request into an ordered execution plan.

Available workers:

CRM:
Can retrieve invoices, customers, orders,
and CRM records.

RAG:
Can search documents and knowledge bases.

MCP:
Can perform external tool/API operations.

Email:
Can send emails and attachments.

Schedule:
Can create, update, and cancel schedules.

Important:

The worker descriptions above are capabilities,
NOT tasks. Never copy them into the plan.

Rules:

1. Create only tasks required by the user's request.
2. Each step must be one concrete action.
3. Assign exactly one worker to each step.
4. If a task requires information that is not yet available,
   create a prerequisite step first.
5. Preserve execution order.
6. Do not create generic tasks such as:
   "Call APIs",
   "Execute external tools",
   "Perform external actions".
7. Do not repeat tasks.
8. Do not execute tasks.
9. Return only the structured plan.

Example:

User:
"Find my invoice and send it by email."

Plan:

Step 1:
agent = CRM
task = "Find the user's invoice."

Step 2:
agent = email
task = "Send the retrieved invoice by email."
"""
    ),
    MessagesPlaceholder("messages"),
])


planner_llm = (
    planner_prompt
    | llm.with_structured_output(Plan)
)

async def planner_node(state: AgentState):

    plan = await planner_llm.ainvoke({
        "messages": state["messages"]
    })

    return {
        "plan": plan,
        "current_step": 0,
        "results": {},
    }


async def crm_worker(state: AgentState):

    step = state["plan"].steps[state["current_step"]]

    print("CRM TASK:", step.task)

    # Your actual CRM agent/tool here
    result = "Invoice INV-123 found."

    results = state.get("results", {}).copy()

    results[state["current_step"]] = result

    return {
        "results": results
    }
async def rag_worker(state: AgentState):

    step = state["plan"].steps[state["current_step"]]

    print("RAG TASK:", step.task)

    # Your actual RAG agent here
    result = "Relevant information retrieved."

    results = state.get("results", {}).copy()

    results[state["current_step"]] = result

    return {
        "results": results
    }

async def mcp_worker(state: AgentState):

    step = state["plan"].steps[state["current_step"]]

    print("MCP TASK:", step.task)

    # Your actual MCP agent/tool here
    result = "MCP operation completed."

    results = state.get("results", {}).copy()

    results[state["current_step"]] = result

    return {
        "results": results
    }

async def email_worker(state: AgentState):

    step = state["plan"].steps[state["current_step"]]

    print("EMAIL TASK:", step.task)

    # Your actual email agent/tool here
    result = "Invoice sent by email."

    results = state.get("results", {}).copy()

    results[state["current_step"]] = result

    return {
        "results": results
    }

async def schedule_worker(state: AgentState):

    step = state["plan"].steps[state["current_step"]]

    print("SCHEDULE TASK:", step.task)

    # Your actual scheduling agent/tool here
    result = "Schedule created."

    results = state.get("results", {}).copy()

    results[state["current_step"]] = result

    return {
        "results": results
    }

def advance(state: AgentState):

    return {
        "current_step": state["current_step"] + 1
    }

def route_next(state: AgentState):

    current_step = state["current_step"]
    plan = state["plan"]

    # All tasks completed
    if current_step >= len(plan.steps):
        return "finish"

    # Get the next worker
    return plan.steps[current_step].agent

def route_supervisor(state: AgentState):

    return state["next"]


builder = StateGraph(AgentState)


# ------------------------------------------------------------
# Nodes
# ------------------------------------------------------------

builder.add_node(
    "supervisor",
    supervisor_node
)

builder.add_node(
    "planner",
    planner_node
)

builder.add_node(
    "crm_worker",
    crm_worker
)

builder.add_node(
    "rag_worker",
    rag_worker
)

builder.add_node(
    "mcp_worker",
    mcp_worker
)

builder.add_node(
    "email_worker",
    email_worker
)

builder.add_node(
    "schedule_worker",
    schedule_worker
)

builder.add_node(
    "advance",
    advance
)


# ------------------------------------------------------------
# START → Supervisor
# ------------------------------------------------------------

builder.add_edge(
    START,
    "supervisor"
)


# ------------------------------------------------------------
# Supervisor → Planner
# ------------------------------------------------------------

builder.add_conditional_edges(
    "supervisor",
    route_supervisor,
    {
        "planner": "planner",
        "finish": END,
    }
)


# ------------------------------------------------------------
# Planner → First Worker
# ------------------------------------------------------------

builder.add_conditional_edges(
    "planner",
    route_next,
    {
        "CRM": "crm_worker",
        "rag": "rag_worker",
        "MCP": "mcp_worker",
        "email": "email_worker",
        "schedule": "schedule_worker",
        "finish": END,
    }
)


# ------------------------------------------------------------
# Worker → Advance
# ------------------------------------------------------------

builder.add_edge(
    "crm_worker",
    "advance"
)

builder.add_edge(
    "rag_worker",
    "advance"
)

builder.add_edge(
    "mcp_worker",
    "advance"
)

builder.add_edge(
    "email_worker",
    "advance"
)

builder.add_edge(
    "schedule_worker",
    "advance"
)


# ------------------------------------------------------------
# Advance → Next Worker / END
# ------------------------------------------------------------

builder.add_conditional_edges(
    "advance",
    route_next,
    {
        "CRM": "crm_worker",
        "rag": "rag_worker",
        "MCP": "mcp_worker",
        "email": "email_worker",
        "schedule": "schedule_worker",
        "finish": END,
    }
)


graph = builder.compile()

```

```python
from pydantic import BaseModel,Field,ConfigDict
from typing_extensions import Literal,Annotated,Dict,Any
from langchain_openai import ChatOpenAI

from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.messages import AIMessage,HumanMessage,SystemMessage,RemoveMessage,ToolMessage,BaseMessage
from langgraph.graph import add_messages,START,END
from dataclasses import dataclass,field

llm = ChatOpenAI(
    model="Qwen2.5-Omni-7B-Q4_K_M.gguf", #multimodel llm with mmproj a1990 mac
    base_url="http://127.0.0.1:8080/v1",
    api_key="not-needed",
    stream_chunk_timeout=0,
    )

@dataclass
class AgentState:    #every key is optional
    messages: Annotated[list[BaseMessage | str | dict[str, Any]],add_messages] =field(default_factory=list)
    plan: Plan | None = None
    next:Literal["__end__","planning"]="planning"
    current_step: int = 0
    results: dict[int, str] = field(default_factory=dict)
    final_result: str | None = None

# class AgentState(TypedDict, total=False):  #With total=False, every key is optional:
#     messages: Annotated[list[BaseMessage | str | dict[str, Any]],add_messages]
#     plan: Plan
#     current_task: int
#     result: dict[int, str]
#     final_result: str
```

```python

class PlanStep(BaseModel):
    model_config=ConfigDict(extra="forbid")
    agent:Literal["rag","mcp","crm","email",'general',"schedule"]
    task:str=Field(min_length=1,description="one concrete task for the selected worker")

class Plan(BaseModel):
    model_config=ConfigDict(extra="forbid")
    steps:List[PlanStep]
planner_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are a Planning Agent.

Your job is ONLY to convert the user's request into an
ordered execution plan.

You do NOT execute any task.

========================
WORKER CAPABILITIES
========================

CRM:
- Retrieve invoices
- Retrieve customers
- Retrieve orders
- Retrieve CRM records


RAG:
- Search documents
- Search knowledge bases
- Retrieve information from indexed documents

MCP:
- Perform external tool/API operations
- Use external systems when no specialized worker is appropriate

Email:
- Send emails
- Send attachments
- Reply to emails

Schedule:
- Create schedules
- Update schedules
- Cancel schedules

General:
- Reasoning, 
-greeting 
-tasks that do not require any specialized worker

========================
MANDATORY ROUTING RULE
========================

Choose the worker based on the ACTUAL ACTION being performed.

The worker must have the capability required to perform the task.

Examples:

"Find my invoice"
-> CRM

"Retrieve my customer record"
-> CRM

"Search the company policy"
-> RAG

"question anser related task"
-> General

"Send the invoice by email"
-> Email

"Create a meeting tomorrow"
-> Schedule

"Call an external API"
-> MCP

NEVER assign invoice retrieval to Email.

NEVER assign email sending to CRM.

NEVER assign document search to Email.

========================
PLANNING RULES
========================

1. Create only tasks required by the user's request.

2. Each step must contain exactly ONE concrete action.

3. Assign exactly ONE worker to each step.

4. If a later task depends on information produced
   by an earlier task, create the prerequisite first.

5. Preserve dependency and execution order.

6. Do not create generic tasks such as:
   - "Call APIs"
   - "Execute external tools"
   - "Perform external actions"

7. Do not repeat tasks.

8. Do not execute tasks.

9. Return ONLY the structured Plan.

========================
EXAMPLE
========================

User:
"Find my invoice and send it by email."

Correct plan:

Step 1:
agent = "crm"
task = "Find the user's invoice."

Step 2:
agent = "email"
task = "Send the retrieved invoice by email."

========================

Another example:

User:
"Find my customer details and email them to me."

Correct plan:

Step 1:
agent = "crm"
task = "Find the user's customer details."

Step 2:
agent = "email"
task = "Send the retrieved customer details by email."

"""),
    MessagesPlaceholder("messages"),
])

planner_llm = planner_prompt | llm.with_structured_output(Plan)
```

```python
class SuperVisorResponse(BaseModel):
    model_config=ConfigDict(extra="forbid")
    next:Literal["planning","__end__"]

supervisor_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are the Supervisor Agent.

Your job is to decide whether the user's request
needs to be planned.

Rules:

1. If the user asks to perform any task, return "planning".
2. Do not directly select workers.
3. Do not execute tasks.
4. Do not choose CRM, RAG, MCP, email, or schedule.
5. Use "finish" only when no task needs to be executed.

Return only the structured response.
"""
    ),
    MessagesPlaceholder("messages"),
])

supervisor_llm = supervisor_prompt | llm.with_structured_output(SuperVisorResponse)

def SuperVisor_node(state:AgentState):
    res = supervisor_llm.invoke({"messages":state.messages})
    return {"next":res.next}
```

```python
def route_next(state:AgentState):      #this is Consditional edge it return node 
    plan = state.plan
    current_step = state.current_step

    if current_step>=len(plan.steps): 
        return END

    return plan.steps[current_step].agent


def planning(state:AgentState)->Literal["rag","mcp",'email','schedule','general','crm']:
    plan = planner_llm.invoke({"messages":state.messages})
    return {"plan":plan}

def supervisor(state:AgentState):
    res = supervisor_llm.invoke({"messages":state.messages})
    return {"next": res.next }

def crm(state:AgentState):
    result = "crm result"
    return {"results": {**state.results,state.current_step: result}}

def mcp(state:AgentState):
    result = "mcp result"
    return {"results": {**state.results,state.current_step: result}}

def email(state:AgentState):
    result = "email result"
    return {"results": {**state.results,state.current_step: result}}

def rag(state:AgentState):
    result = "rag result"
    return {"results": {**state.results,state.current_step: result}}

def schedule(state:AgentState):
    result = "schedule result"
    return {"results": {**state.results,state.current_step: result}}

def general(state:AgentState):
    result = "general result"
    return {"results": {**state.results,state.current_step: result}}

def advance(state:AgentState):
    return {"current_step":state.current_step +1}

def route_supervisor(state: AgentState):

    return state.next


```

```python
graph =(StateGraph(state_schema=AgentState)
        .set_entry_point("supervisor")
        .add_node("supervisor",supervisor)
        .add_node("planning",planning)
        .add_node("advance",advance)
        .add_node("crm",crm)
        .add_node("rag",rag)
        .add_node("email",email)
        .add_node("mcp",mcp)
        .add_node("schedule",schedule)
        .add_node("general",general)
        .add_conditional_edges( "supervisor",route_supervisor,{"__end__":END,"planning":'planning'} )
        .add_conditional_edges("planning",route_next,['rag',"crm",'mcp','email','schedule','general',END])
        .add_conditional_edges("advance",route_next,['rag',"crm",'mcp','email','schedule','general',END])
        .add_edge('mcp','advance')
        .add_edge('rag','advance')
        .add_edge('email','advance')
        .add_edge('crm','advance')
        .add_edge('general','advance')
        .add_edge('schedule','advance')
        .compile() )

        
```
